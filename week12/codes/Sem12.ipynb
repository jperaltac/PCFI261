```python
# -- tiempos --
# 105 -> 06:14
# 136 -> 06:45
# dt = 310 ms -> 0.31 segundos
# -- largos --
# 105 -> 5ft
# 136 -> 0ft
# dy = 5ft -> 1.524 metros
# 5ft = 658 pixeles
# -> 1 px = 1.524 / 658 metros
y = puntos[:,1][105:136] * 1.524 / 658
t = np.linspace(0, (136-105)/60, len(y))
print(y)
```

```python
plt.plot(t, y, label="t vs y")
g, v0, h = np.polyfit(t, y, 2)
plt.plot(t, np.polyval([g, v0, h], t), label=f"g={g:.2f}")
plt.plot(t, np.polyval([9.8, 0, h], t), label=f"g=9.8")
plt.legend()
print(f"{g:.2f}t² + {v0:.2f}t + {h:.2f}")
```

```python
from sklearn.model_selection import train_test_split
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
t_train, t_val, y_train, y_val = train_test_split(
    t, y, test_size=0.25, shuffle=True, random_state=123
)
```

```python
tf.keras.utils.set_random_seed(123)

model = keras.Sequential([
    layers.Input(shape=(1,)),
    layers.Dense(32, activation="tanh"),
    layers.Dropout(0.2),
    layers.Dense(32, activation="tanh"),
    layers.Dense(1)
])
```

```python
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.01),
    loss=keras.losses.MeanSquaredError(),
)
```

```python
history = model.fit(
    t_train, y_train,
    validation_data=(t_val, y_val),
    epochs=300,
)
```

```python
t_grid = np.linspace(np.min(t), np.max(t), 500)
y_pred = model.predict(t_grid, verbose=0)
```

```python
plt.plot(t, y, "b.", label="valores reales")
plt.plot(t_grid, y_pred, "r.", label="valores predecidos")
plt.xlabel("tiempo")
plt.ylabel("altura")
plt.legend()
```

$$\sum_i F_i = m \frac{d²y}{dt²}$$

$$-g = \frac{d²y}{dt²}$$




```python
def derivadas_y(model, t_in):
    with tf.GradientTape() as tape2:
        tape2.watch(t_in)
        with tf.GradientTape() as tape1:
            tape1.watch(t_in)
            y = model(t_in)

        dy_dt = tape1.gradient(y, t_in)
        d2y_dt2 = tape2.gradient(dy_dt, t_in)
    return y, dy_dt, d2y_dt2

t_phys = tf.convert_to_tensor(t_grid, dtype=tf.float32)
y_p, dy_p, d2y_p = derivadas_y(model, t_phys)
```

```python
plt.plot(t_phys, y_p)
plt.plot(t_phys, dy_p)
plt.plot(t_phys, d2y_p)
```


$$\frac{d²y}{dt²} + g = 0$$
$$\frac{d^2y}{dt^2} \frac{m}{s^2} + g = 0$$


```python
g = 9.8
residual = d2y_p + g
loss_phys = tf.reduce_mean(tf.square(residual))
print(loss_phys)
```

```python
g_raw = tf.Variable(np.log(g), dtype=tf.float32)
g_param = tf.nn.softplus(g_raw)
```
